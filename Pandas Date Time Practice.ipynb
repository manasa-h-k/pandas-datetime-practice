Q50) Find the longest streak of consecutive days with orders above 3.
<details>
<summary>Show Solution</summary>

```python
daily_orders = df.resample("D", on="date")["order_id"].count()
mask = daily_orders > 3
streaks = (mask != mask.shift()).cumsum()
longest = mask.groupby(streaks).sum().max()

```
